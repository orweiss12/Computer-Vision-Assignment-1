**Inline Question 2**

Although gradcheck is reliable softmax loss, it is possible that for SVM loss, once in a while, a dimension in the gradcheck will not match exactly. What could such a discrepancy be caused by? Is it a reason for concern? What is a simple example in one dimension where a svm loss gradient check could fail? How would change the margin affect of the frequency of this happening?

Note that SVM loss for a sample $(x_i, y_i)$ is defined as: $$L_i = \sum_{j\ne y_i}\max(0, s_j - s_{y_i} + \Delta)$$ Where $j$ iterates over all classes except the correct class $y_i$ and $s_j$ denotes the classifier score for $j^{th}$ class. $\Delta$ is a scalar margin.
*Hint: the SVM loss function is not strictly speaking differentiable.*


$\color{blue}{\textit Your Answer:}$ The discrepancy during gradcheck for SVM loss occurs because the function is not strictly differentiable at the "knee" where the value inside the max function equals exactly zero ($s_j - s_{y_i} + \Delta = 0$). Because the numerical gradient check evaluates the function using a small finite step $h$, this interval can inadvertently cross the non-differentiable threshold, yielding a numerical slope that differs from the exact analytical subgradient. This is not a cause for concern, as the probability of landing precisely on this specific point during optimization is exceedingly small, and subgradient descent handles it reliably. A simple 1D example is $f(x) = \max(0, x)$: if we evaluate it at $x = -0.001$ with a numerical step of $h = 0.01$, the analytical gradient is $0$ (since $x < 0$), but the numerical gradient will be positive because the step crosses into the $x > 0$ region. Finally, changing the scalar margin $\Delta$ will not affect the frequency of these mismatches; it merely shifts the location of the non-differentiable point along the axis, and the model's weights will simply scale to adapt to the new margin during training.

$\color{blue}{\textit Your Answer:}$ The visualized Softmax classifier weights appear as blurry, generalized templates or "prototypes" of the 10 object classes in the CIFAR-10 dataset (e.g., a car, a frog, a horse).

This happens because the linear Softmax classifier learns a single row of weights (a template) for each class. During training, the classifier uses the dot product to compute how similar an input image is to each class's template. To minimize the loss across all variations of a single class in the training data, the optimizer averages these variations into one composite image. For instance, if cars in the dataset are both red and blue and face different directions, the resulting "car" weight template will be a blurry amalgamation of these colors and orientations.

$\color{blue}{\textit Your Answer:}$ true


$\color{blue}{\textit Your Explanation:}$ SVM loss (Hinge Loss): This function calculates a penalty only when an example violates the margin limits (Margin). If we add a new example that is correctly classified with high confidence (ie the difference between the score of the correct class and the score of the other classes is greater than the defined margin interval), the SVM loss for it will be exactly zero. Adding a zero to the amount will leave the total loss unchanged.

Softmax Loss (Cross-Entropy): This function never reaches exactly zero, because it requires the probability of the correct class to be exactly 1 (which happens only when the score approaches infinity). Therefore, any new point added to the training set - even one that is perfectly classified - will always add some positive value greater than zero. This addition will necessarily change the total Softmax loss amount.
